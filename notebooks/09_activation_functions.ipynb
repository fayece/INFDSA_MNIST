```python
from infdsa_mnist import mnist_load_normalized, mnist_output, activation_functions, activation_functions_output
import numpy as np
```

In the previous chapter, we have seen how to build a simple neural network. We let Keras handle the activation functions for us. In this chapter, we will zoom in on the activation functions and implement them ourselves.

```python
(train_data, test_data) = mnist_load_normalized.load_mnist_flattened_normalized()
x_train, y_train = train_data
x_test, y_test = test_data
```

### Making pre-activation explicit

```python
x = x_train[:10]

num_features = 784
num_neurons = 32

w = np.random.randn(num_features, num_neurons) * 0.01

b = np.zeros(num_neurons)

z = activation_functions.pre_activation(x, w, b)
print(z.shape)
print(z)
```

The value of the pre-activation `z` receives the shape of (10, 32), which corresponds to the 10 samples in the batch and the 32 neurons in the layer. Each element in `z` represents the weighted sum of the input features for a particular neuron before applying the activation function. The values in `z` can be positive, negative, or zero, depending on the weights and inputs.

`z` is not yet the output of a neuron, as it has not been passed through an activation function. This is merely a step in the process to compute the output of the neurons. An activation function will be applied to `z` to let the network learn complex patterns in the data.

### ReLU

```python
relu_int_neg = activation_functions.relu(-1)
print(relu_int_neg)

relu_int_nil = activation_functions.relu(0)
print(relu_int_nil)

relu_int_pos = activation_functions.relu(1)
print(relu_int_pos)

relu_arr = activation_functions.relu(z)
print(relu_arr)
```

### Sigmoid

```python
sigmoid_int_neg = activation_functions.sigmoid(-1_000_000)
print(sigmoid_int_neg)

sigmoid_int_nil = activation_functions.sigmoid(0)
print(sigmoid_int_nil)

sigmoid_int_pos = activation_functions.sigmoid(1_000_000)
print(sigmoid_int_pos)
```

The sigmoid function ($\sigma(z) = \frac{1}{1 + e^{-z}}$) maps input values to a range between 0 and 1. For large negative inputs, the sigmoid function approaches 0, while for large positive inputs, it approaches 1. For an input of 0, the sigmoid function outputs 0.5, as an input of 0 makes sigmoid perfectly unsure whether the input is positive or negative. This behavior makes the sigmoid function useful for binary classification problems, where the output can be interpreted as a probability of belonging to a particular class.

As we can see from the outputs, `-1_000_000` and `1_000_000` are extremely large values for the sigmoid function. Euler's number `e` raised to the power of `-1_000_000` cannot be represented properly and results in an underflow, which is why we get 0 for the negative output. On the other hand, `e` raised to the power of `1_000_000` results in an overflow, which is why we get 1 for the positive output.

Let's show some "valid" values for the sigmoid function to see how it behaves in a more typical range.

```python
sigmoid_int_neg = activation_functions.sigmoid(-9)
print(sigmoid_int_neg)

sigmoid_int_pos = activation_functions.sigmoid(9)
print(sigmoid_int_pos)
```

These values are more typical for the sigmoid function. The negative and positive show the expected behavior of negative numbers being mapped close to 0 and positive numbers being mapped close to 1.

### Softmax

```python
z1 = [3.0, 1.0, 0.2]
z2 = [2.5, 4.9, 0.06]
z3 = [0.1, 0.2, 0.3]
z4 = [0.0, 0.0, 0.0]

softmax_z1 = activation_functions.softmax(z1)
print(softmax_z1)
print(np.sum(softmax_z1))

softmax_z2 = activation_functions.softmax(z2)
print(softmax_z2)
print(np.sum(softmax_z2))

softmax_z3 = activation_functions.softmax(z3)
print(softmax_z3)
print(np.sum(softmax_z3))

softmax_z4 = activation_functions.softmax(z4)
print(softmax_z4)
print(np.sum(softmax_z4))
```

The softmax function ($\sigma(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$) converts a vector of raw scores (called logits) into probabilities. The output of the softmax function is a probability distribution over the classes, where each value is between 0 and 1, and the sum of all values is equal to 1 (or very close to 1 due to floating-point precision limitations). In the examples provided, we can see that the softmax function has transformed the input vectors `z1`, `z2`, `z3`, and `z4` into probability distributions. The values in each output vector sum to 1, confirming that they represent valid probability distributions. The softmax function is commonly used in the output layer of neural networks for multi-class classification problems, where it helps to interpret the output as probabilities for each class. In a multi-class classification problem, the class with the highest probability can be selected as the predicted class for a given input.

### Neural Network with self-implemented activation functions

Before we can implement a neural network with self-implemented activation functions, let's create a simple neural network without activation functions to see what the output looks like. We will then add the activation functions to see how they affect the output.

```python
model_raw = activation_functions.create_nn_no_activation(x_train.shape[1])
activation_functions.compile_nn(model_raw)
activation_functions.fit_nn(model_raw, x_train, y_train, epochs=2, batch_size=16, verbose=True)
```

```python
output = activation_functions_output.get_hidden_layer_output(model_raw, x_test)
print(output.shape)
print(output)
```

The output of the hidden layer without activation functions is a linear transformation of the input data. The values can be positive, negative, or zero. In order for the neural network to learn complex patterns in the data, we need to introduce activation functions. Let's add our self-implemented activation functions to the hidden layer and see how the output changes. Do note that we will use slightly different activation functions than the ones we implemented earlier, as we used NumPy for earlier implementations, and Keras does not like us to use NumPy functions in the model. We will recreate the activation functions using Keras instead.

```python
model = activation_functions.create_nn(x_train.shape[1])
activation_functions.compile_nn(model)
activation_functions.fit_nn(model, x_train, y_train, epochs=2, batch_size=16, verbose=True)
```

```python
output = activation_functions_output.get_hidden_layer_output(model, x_test)
print(output.shape)
print(output)
```

As we can see, the output of the hidden layer with activation functions no longer features negative values. The activation functions have introduced non-linearity to the model, allowing it to learn more complex patterns in the data. Do, however, note that the numbers printed are not entirely the same as the ones within the model itself. This is because the `get_hidden_layer_output` function does not take the Batch Normalization layer into account, which is present in the model. The Batch Normalization layer normalizes the output of the hidden layer, which can affect the values we see when we print the output. Since both models above do feature Batch Normalization, this should not pose a problem for our comparison. The main takeaway is that the activation functions have introduced the ability to learn complex patterns in the data, which is essential for performance.

### Output layer activation functions

To end off this chapter, let's take a look at the output layer activation functions. Typically, we would use the softmax activation function for multi-class classification problems, as it converts the raw output scores into probabilities. However, we can use other activation functions as well, such as sigmoid. Let's see how the output changes between using softmax and sigmoid in the output layer. Considering the model with activation functions already has a softmax activation function in the output layer, we will use the raw model without activation functions to test the effect of different output layer activation functions. Our aim with this exercise will not be to achieve good performance, but rather to see the effect of different activation functions on the output.

```python
raw_model_output = model_raw.predict(x_test[:1])
print(f"Raw final layer output (Logits):\n {raw_model_output}")

raw_model_output_sigmoid = activation_functions.sigmoid(raw_model_output)
print(f"\nSigmoid output:\n {raw_model_output_sigmoid}")

raw_model_output_softmax = activation_functions.softmax(raw_model_output)
print(f"\nSoftmax output:\n {raw_model_output_softmax}")
```

Let's go through these results.

The raw final layer output (logits) is a vector of raw scores for each class. These scores can be positive, negative, or zero, and they do not represent probabilities. However, we can tell by the values that the model is more confident in some classes than others, as some scores are higher than others. From this, we can make a guess about which class the model would predict, which would be the class with the highest score.

When we apply the sigmoid function to the raw output, we get values between zero and one for each class. However, sigmoid treats every value as a separate yes/no question, as opposed to treating all values as part of a whole. This means that the output does not represent a valid probability distribution, as the values do not sum to 1. It still gives us an idea of how confident the model is in each class, but it does not allow us to directly compare the classes against each other.

When we apply the softmax function to the raw output, we once again get values between zero and one for each class. In contrast to sigmoid, softmax treats the output as a whole and converts it into a valid probability distribution. The values sum to 1, which allows us to directly compare the classes against each other. The class with the highest probability is the one that the model would predict.
